## 1. Jupyterとは？

これまでのPythonの授業では、

1. エディターにプログラムを書く
2. `.py` ファイルとして保存する
3. ターミナルからプログラムを実行する
4. 実行結果を確認する

という方法を使ってきました。

この授業では、**Jupyter**を使ってPythonを実行します。

Jupyterでは、プログラムを「セル」と呼ばれる単位に分けて記述し、セルごとに実行できます。実行結果やグラフも、そのセルのすぐ下に表示されます。

| | これまでの方法 | Jupyter |
|---|---|---|
| コードを書く | エディター | Codeセル |
| 保存 | `.py` ファイル | `.ipynb` ファイル |
| 実行 | ターミナルから実行 | セルごとに実行 |
| 結果 | ターミナルなどで確認 | Notebook内で確認 |

Jupyterは、コードを少しずつ実行して結果を確認できるため、データ解析のように試行錯誤しながら進める作業に便利です。

一方、大きなプログラムの作成や繰り返し実行には、通常の `.py` ファイルの方が適している場合もあります。

## 2. Notebook形式でPythonを実行できる主な環境

Pythonのコード、説明文、実行結果、グラフなどを1つのNotebook上で扱える環境には、いくつかの種類があります。

| 環境 | 特徴 |
|---|---|
| **Jupyter Notebook / JupyterLab** | PCにPython・Jupyter環境を構築して利用。自由度が高い一方、環境構築が必要。 |
| **JupyterLite** | ブラウザ上で動作。インストール不要。ただし、一部の機能やライブラリに制限。 |
| **Google Colaboratory** | Googleが提供するクラウド上のNotebook環境。Googleアカウントが必要。 |

この授業では、環境構築を行わず、全員が共通した環境をすぐに利用できるようにするため、**JupyterLite**を使用します。

## 5. Notebookの保存について

JupyterLiteで編集したNotebookの作業内容は、使用しているブラウザ内に保存されます。

同じPC・同じブラウザであれば、後から作業を続けることができます。

ただし、次の場合には作業内容が引き継がれないことがあります。

- 別のPCを使う
- 別のブラウザを使う
- ブラウザのデータを削除する

### NotebookをPCにダウンロードする

Notebookを提出するときは、作業したNotebookを `.ipynb` ファイルとしてPCにダウンロードします。

1. 画面上部のメニューから **File** をクリックします。
2. **Save and Export Notebook As** を選択します。
3. **Notebook (.ipynb)** をクリックします。
4. `.ipynb` ファイルがPCにダウンロードされたことを確認します。

ダウンロードした `.ipynb` ファイルをMoodleから提出してください。

**提出前に、作業した内容が保存された `.ipynb` ファイルをダウンロードできていることを必ず確認してください。**

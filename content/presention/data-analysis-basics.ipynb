## 2. NumPyで数値データを扱う

Pythonのリストでも複数の値を扱えますが、数値データの計算にはNumPyの配列が便利です。

まず、5日分の水温をNumPy配列として作ってみます。

````python
temperature = np.array([15.2, 15.5, 15.1, 15.8, 16.0])

temperature
````

NumPy配列では、すべての要素に対してまとめて計算できます。

例えば、水温をすべて1℃上げると、

````python
temperature + 1
````

同様に、平均値や最大値、最小値なども簡単に計算できます。

````python
print("平均:", np.mean(temperature))
print("最大:", np.max(temperature))
print("最小:", np.min(temperature))
````

## 3. pandasで表形式のデータを扱う

実際の観測データには、水温だけでなく、日付や塩分など複数の情報が含まれることがあります。

pandasでは、このような表形式のデータを **DataFrame（データフレーム）** として扱います。

````python
data = pd.DataFrame({
    "date": ["2025-01-01", "2025-01-02", "2025-01-03", "2025-01-04", "2025-01-05"],
    "temperature": [15.2, 15.5, 15.1, 15.8, 16.0],
    "salinity": [33.8, 33.7, 33.9, 33.6, 33.5]
})

data
````

DataFrameでは、列名を使って必要なデータを取り出せます。

````python
data["temperature"]
````

複数の列を取り出す場合は、列名をリストで指定します。

````python
data[["temperature", "salinity"]]
````

## 4. CSVファイルを読み込む

実際のデータ解析では、データがCSVファイルとして保存されていることがよくあります。

pandasの `read_csv()` を使うと、CSVファイルをDataFrameとして読み込めます。

ここでは、同じフォルダにある `sample_temperature.csv` を読み込みます。

````python
df = pd.read_csv("sample_temperature.csv")

df
````

データが大きい場合は、最初の数行だけを確認すると便利です。

````python
df.head()
````

列名は `columns` で確認できます。

````python
df.columns
````

データの行数・列数は `shape` で確認できます。

````python
df.shape
````

各列の基本的な統計量は `describe()` で確認できます。

````python
df.describe()
````

### 列の平均値を計算する

DataFrameから列を取り出し、`mean()` を使うと、その列の平均値を計算できます。

例えば、水温 `temperature` の平均値は次のように計算できます。

````python
df["temperature"].mean()
````

同じように、`max()` や `min()` を使うと、最大値や最小値を計算できます。

````python
print("平均:", df["temperature"].mean())
print("最大:", df["temperature"].max())
print("最小:", df["temperature"].min())
````

## 5. 日付を扱う

CSVから読み込んだ直後の `date` は、文字列として扱われています。

時系列データを解析するときは、日付として扱える形式に変換しておくと便利です。

````python
df["date"] = pd.to_datetime(df["date"])

df.head()
````

データ型は `dtypes` で確認できます。

````python
df.dtypes
````

## 6. matplotlibでグラフを描く

最後に、水温の時間変化をグラフにしてみます。

横軸を日付、縦軸を水温として折れ線グラフを描きます。

````python
plt.plot(df["date"], df["temperature"])

plt.xlabel("Date")
plt.ylabel("Temperature")
plt.show()
````

点の位置を分かりやすくするため、`marker="o"` を追加することもできます。

````python
plt.plot(df["date"], df["temperature"], marker="o")

plt.xlabel("Date")
plt.ylabel("Temperature")
plt.show()
````

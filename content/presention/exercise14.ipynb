### 課題4のヒント

年を平均からの偏差 $x_i'$ とし、

$$
\sin\left(\frac{2\pi x_i'}{22}\right),
\qquad
\cos\left(\frac{2\pi x_i'}{22}\right)
$$

を作ります。

係数は、

$$
A=
\frac{\sum_i f_i\sin(2\pi x_i'/22)}
{\sum_i\sin^2(2\pi x_i'/22)},
\qquad
B=
\frac{\sum_i f_i\cos(2\pi x_i'/22)}
{\sum_i\cos^2(2\pi x_i'/22)}
$$

です。

22年周期の変動は

$$
A\sin\left(\frac{2\pi x_i'}{22}\right)
+
B\cos\left(\frac{2\pi x_i'}{22}\right)
$$

で、振幅は

$$
\sqrt{A^2+B^2}
$$
